## Leitura do arquivo | Tamanho do arquivo

```python
# Bibliotecas
import pandas as pd  # Leitura e manipulação dos arquivos
import os  # Manipulação de caminhos

# Local do arquivo (caminho relativo ao projeto)
caminho = "../data/raw/cancelamentos.csv"

# Tamanho do arquivo em disco (MB)
print(f"Tamanho do arquivo em disco: {os.path.getsize(caminho) / 1024 ** 2:.2f} MB")

# Leitura do arquivo (usando o Pandas)
df_raw = pd.read_csv(
    caminho,
    sep=",",
    encoding="utf-8",
    decimal=".",
    low_memory=False,
)

# Total de linha e colunas
print(f"Linhas: {df_raw.shape[0]}, Colunas: {df_raw.shape[1]}")

# Tamanho do dataframe na memória (MB)
print(
    f"Tamanho do dataframe na memória: {df_raw.memory_usage(deep=True).sum() / 1024 ** 2:.2f} MB"
)
```
