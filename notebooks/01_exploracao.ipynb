## Leitura do arquivo | Tamanho do arquivo

```python
# Bibliotecas
import pandas as pd  # Leitura e manipulação de arquivos
import os  # Informações do arquivo no disco

# Local do arquivo
caminho = "../data/raw/cancelamentos.csv"

# Leitura do arquivo
df_raw = pd.read_csv(
    caminho,
    sep=",",
    encoding="utf-8",
    decimal=".",
    low_memory=False,
)
```

## Inicio do dataframe

```python
# Ínicio do dataframe
display(df_raw.head())
```

## Final do dataframe

```python
# Final do dataframe
display(df_raw.tail())
```

## Meio do dataframe

```python
# Meio do dataframe
display(df_raw.sample(5, random_state=42))
```

**Observação:**
* Início, meio e fim têm o mesmo formato e as mesmas 12 colunas.
* Valores inteiros aparacem com .0 (ex.: CustomerID = 349936.0, idade = 23.0).

## Tipo dos dados

```python
# Tipo de dados e valores não nulos
df_raw.info()

# Tamanho do arquivo em disco
print(f"Tamanho do arquivo no disco: {os.path.getsize("../data/raw/cancelamentos.csv") / 1024**2:.2f} MB")

# Tamanho do arquivo na memória
print(f'Tamanho do dataframe na memória: {df_raw.memory_usage(deep=True).sum() / 1024**2:.2f} MB')
```

**Observação:**
* 50.000 linhas × 12 colunas.
* 3 coluna de texto (`str`): `sexo`, `assinatura` e `duração do contrato`.
* 9 colunas numéricas (`float64`): `CustomerID`, `idade`, `tempo_como_cliente`, `frequencia_uso`, `ligacoes_callcenter`, `dias_atraso`, `total_gasto`, `meses_ultima_interacao`, `cancelou`.
* Nulos: `sexo` (3) e `tempo_como_cliente` (2).
* Colunas que parecem inteiras: (CustomerID, idade, tempo_como_cliente, frequencia_uso, ligacoes_callcenter, dias_atraso, meses_ultima_interacao, cancelou).
* 3,34 MB em disco e 11,35 MB em memória (`deep=True`): a tabela carregada é maior que o arquivo, o que é esperado.

## Valores únicos e repetidos

```python
# Identificador: o ID é único?
print(f"ID único: {df_raw['CustomerID'].nunique() == len(df_raw)}")

# IDs repetidos?
print(f"Quantidad de IDs repetidos: {df_raw['CustomerID'].duplicated().sum()}")

# Variável alvo: quais valores existem?
print(f"Valores da coluna alvo: {df_raw['cancelou'].unique()}")
```

**Observação:**
* CustomerID **não é único:** 1.469 IDs repetidos -> granularidade "1 linha = 1 cliente" não confirmada (investigar na Etapa 3).
* Variável alvo `cancelou` existe, com valores `0.0` e `1.0`.
