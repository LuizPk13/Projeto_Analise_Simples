# Analisando a base de dados de cancelamentos

## Case - Cancelamento de Clientes

Você foi contratado por uma empresa com mas de 800 mil clientes para um projeto de Dados. Recentemente a empresa percebeu que da sua base total de clientes, a maioria são clientes inativos ou seja, que já cancelaram o serviço.

Precisando melhorar seus resultados ela quer conseguir entender os principais motivos desses cancelamentos e quais as ações mais eficientes para reduzir esse número.

```python
# Importando a biblioteca para leitura e manipulação dos dados.
import pandas as pd

# Local do meu arquivo
base = r"C:\Users\luizhomem\Downloads\Aula_02 - Analise de Dados\data\cancelamentos.csv"

# Leitura do DataFrame
df_original = pd.read_csv(
    base,  # Local da base
    sep=",",  # Separador de colunas
    encoding="utf-8",  # Formato de caracteres
    decimal=".",  # Decimal
    low_memory=False,  # Lê o arquivo inteiro e deixa o DataFrame em memória
)

# Cópia do DataFrame (para não alterar a base original)
df = df_original.copy()
```

```python
# Informações da base de dados
display(df.info())

# Visualização dos dados
display(df)
```

```python
# Removendo a coluna CustomerID, pois não é relevante para a análise
df = df.drop(columns=["CustomerID"])
```

```python
# Removendo as linha que possuem valores nulos
df =df.dropna().reset_index(drop=True)
```

```python
display(df.info())
```

# Pergunta principal: Qual é a quantidade de cancelamentos?

```python
# Convertendo a coluna "cancelou" de números 0.0 1.0 para texto
df["cancelou"] = df["cancelou"].map({0.0: "Não", 1.0: "Sim"})

print(f'Quantidade total de clientes:\n{len(df)}')

print(
    f"\nQuantidade de cancelamentos:\n{df['cancelou']
    .value_counts()
    .to_string()}"
)

print(
    f"\nPercentual de cancelamentos:\n{df['cancelou']
    .value_counts(normalize=True)
    .map(lambda x: f'{x * 100:.2f}%')
    .to_string()}"
)
```

```python
# Importando a biblioteca Plotly
import plotly.express as px

fig = px.histogram(
    df,
    x="cancelou",
    color="cancelou",
    template="plotly_white",
    title="Distribuição de cancelamentos",
    color_discrete_sequence=["#2b5c8f", "#e76f51"],
)

fig.update_layout(
    title_x=0.5,
    yaxis_title="Número de cancelamentos",
    xaxis_title="Status do cancelamento",
)

fig.show()
```

# Aonde estão os maiores cancelamentos em cada coluna?

```python
# Colunas que eu identifiquei após uma breve consulta em graficos que chamam a atenção
lista_colunas = [
    "idade",
    "sexo",
    "ligacoes_callcenter",
    "dias_atraso",
    "duracao_contrato",
    "total_gasto"
]

# Percorrendo cada coluna e gerando uma grafico de histograma para cada coluna
for coluna in lista_colunas:
    fig = px.histogram(
        df,
        x=coluna,
        color="cancelou",
        template="plotly_white",
        title="Distribuição de cancelamentos",
        color_discrete_sequence=["#2b5c8f", "#e76f51"],
    )

    # Adiciona os números nas barras
    fig.update_traces(
        texttemplate="%{y}",  # Mostra o valor do eixo y (contagem)
        textposition="inside",  # Posição do texto: 'inside' (dentro), 'outside' (fora) ou 'auto'
    )

    fig.update_layout(
        title_x=0.5,
        yaxis_title="Número de cancelamentos",
    )

    fig.show()
```

# Analise detalhada sobre o gráfico de sexo

```python
# quantidade de clientes ao total
print(f"Quantidade de clientes ao total: {len(df)}")

# Total de clientes homens e mulheres
print(f"\nTotal de cliente por sexo:\n{df['sexo'].value_counts()
    .to_string()}"
)

# Agrupando sexo e cancelamento
sexo_cancelamento = df.groupby('sexo')['cancelou'].value_counts(normalize=True).map('{:.2%}'.format)
display(sexo_cancelamento)
```
