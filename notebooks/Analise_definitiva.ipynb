# Analisando a base de dados de cancelamentos

## Case - Cancelamento de Clientes

Você foi contratado por uma empresa com mas de 800 mil clientes para um projeto de Dados. Recentemente a empresa percebeu que da sua base total de clientes, a maioria são clienetes inativosm ou seja, que já cancelaram o serviço.

Precisando melhorar seus resultados ela quer conseguir entender os principais motivos desses cancelamentos e quais as ações mais eficientes para reduzir esse número.

```python
# Importando a biblioteca para leitura e manipulação dos dados.
import pandas as pd

# Local do meu arquivo
base = r"C:\Users\luizhomem\Downloads\Aula_02 - Analise de Dados\data\cancelamentos.csv"

# Leitura do DataFrame
df_original = pd.read_csv(
    base,  # Local da base
    sep=",",  # Separador de colunas
    encoding="utf-8",  # Formato de caracteres
    decimal=".",  # Decimal
    low_memory=False,  # Lê o arquivo inteiro e deixa o DataFrame em memória
)

# Cópia do DataFrame (para não alterar a base original)
df = df_original.copy()
```

```python
# Informações da base de dados
display(df.info())

# Visualização dos dados
display(df)
```

```python
# Removendo a coluna CustomerID, pois não é relevante para a análise
df = df.drop(columns=["CustomerID"])
```

```python
# Removendo as linha que possuem valores nulos
df =df.dropna().reset_index(drop=True)
```
